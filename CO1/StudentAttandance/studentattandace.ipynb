```python
students = []

def add_student():
    print("\n--- Add Student ---")
    name = input("Enter student name: ").strip()

    if not name:
        print("Invalid student name.")
        return

    try:
        total_classes = int(input("Enter total classes conducted: "))
        attended_classes = int(input("Enter classes attended: "))
    except ValueError:
        print("Please enter valid numbers.")
        return

    if total_classes <= 0:
        print("Total classes must be greater than zero.")
        return

    if attended_classes < 0 or attended_classes > total_classes:
        print("Attended classes must be between 0 and total classes.")
        return

    percentage = (attended_classes / total_classes) * 100

    student = {
        "name": name,
        "total_classes": total_classes,
        "attended_classes": attended_classes,
        "percentage": percentage
    }

    students.append(student)

    print(f"Student {name} added successfully.")
    print(f"Attendance: {percentage:.2f}%")


def display_students():
    print("\n--- Student Attendance Records ---")

    if not students:
        print("No student records available.")
        return

    print(f"{'Name':<20}{'Total':<10}{'Attended':<12}{'Attendance':<12}")
    print("-" * 54)

    for student in students:
        print(
            f"{student['name']:<20}"
            f"{student['total_classes']:<10}"
            f"{student['attended_classes']:<12}"
            f"{student['percentage']:.2f}%"
        )


def students_below_75():
    print("\n--- Students Below 75% Attendance ---")

    below = [student for student in students if student["percentage"] < 75]

    if not below:
        print("No students are below 75% attendance.")
        return

    for student in below:
        print(f"{student['name']} - {student['percentage']:.2f}%")


def highest_attendance():
    print("\n--- Highest Attendance ---")

    if not students:
        print("No student records available.")
        return

    highest = max(student["percentage"] for student in students)

    for student in students:
        if student["percentage"] == highest:
            print(f"{student['name']} - {student['percentage']:.2f}%")


def class_average():
    print("\n--- Class Average Attendance ---")

    if not students:
        print("No student records available.")
        return

    average = sum(student["percentage"] for student in students) / len(students)
        
    print(f"Class average attendance: {average:.2f}%")


def main():
    while True:
        print("\n--- Student Attendance Management System ---")
        print("1. Add Student")
        print("2. Display All Students")
        print("3. Students Below 75%")
        print("4. Highest Attendance")
        print("5. Class Average Attendance")
        print("6. Exit Program")

        choice = input("Enter choice: ").strip()

        if choice == "1":
            add_student()
        elif choice == "2":
            display_students()
        elif choice == "3":
            students_below_75()
        elif choice == "4":
            highest_attendance()
        elif choice == "5":
            class_average()
        elif choice == "6":
            print("Exiting system.")
            break
        else:
            print("Invalid choice, try again.")


if __name__ == "__main__":
    main()
```
