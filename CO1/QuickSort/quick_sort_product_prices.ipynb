```python
def quick_sort(arr):
    if len(arr) <= 1:
        return arr

    pivot = arr[len(arr) // 2]

    left = [x for x in arr if x < pivot]
    middle = [x for x in arr if x == pivot]
    right = [x for x in arr if x > pivot]

    return quick_sort(left) + middle + quick_sort(right)


products = [850, 450, 12500, 250, 1800, 3200, 1400, 2600]

print("Original product prices:", products)

sorted_prices = quick_sort(products)

print("Sorted product prices:", sorted_prices)
```
