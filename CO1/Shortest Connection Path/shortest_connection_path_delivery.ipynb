```python
import heapq


def shortest_path(graph, start, destination):
    distances = {location: float('inf') for location in graph}
    previous = {location: None for location in graph}

    distances[start] = 0
    priority_queue = [(0, start)]

    while priority_queue:
        current_distance, current_location = heapq.heappop(priority_queue)

        if current_distance > distances[current_location]:
            continue

        if current_location == destination:
            break

        for neighbor, distance in graph[current_location].items():
            new_distance = current_distance + distance

            if new_distance < distances[neighbor]:
                distances[neighbor] = new_distance
                previous[neighbor] = current_location
                heapq.heappush(
                    priority_queue,
                    (new_distance, neighbor)
                )

    if distances[destination] == float('inf'):
        return None, float('inf')

    path = []
    current = destination

    while current is not None:
        path.append(current)
        current = previous[current]

    path.reverse()

    return path, distances[destination]


graph = {
    "Vijayawada": {
        "Guntur": 32,
        "Amaravati": 18,
        "Machilipatnam": 65
    },
    "Guntur": {
        "Vijayawada": 32,
        "Tenali": 30,
        "Narasaraopet": 55
    },
    "Amaravati": {
        "Vijayawada": 18,
        "Guntur": 35,
        "Mangalagiri": 12
    },
    "Machilipatnam": {
        "Vijayawada": 65,
        "Gudivada": 45
    },
    "Tenali": {
        "Guntur": 30,
        "Mangalagiri": 28
    },
    "Narasaraopet": {
        "Guntur": 55
    },
    "Mangalagiri": {
        "Amaravati": 12,
        "Tenali": 28
    },
    "Gudivada": {
        "Machilipatnam": 45
    }
}


print("Available locations:")
print(", ".join(graph.keys()))

start = input("Enter starting location: ").strip().title()
destination = input("Enter destination: ").strip().title()

if start not in graph or destination not in graph:
    print("Invalid location.")
elif start == destination:
    print("Starting location and destination are the same.")
else:
    path, distance = shortest_path(graph, start, destination)

    if path is None:
        print("No connection found.")
    else:
        print("\nShortest connection path:")
        print(" -> ".join(path))
        print(f"Total distance: {distance} km")
```
