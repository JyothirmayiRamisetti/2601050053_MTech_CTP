```python
def bubble_sort(arr):
    n = len(arr)

    for i in range(n):
        swapped = False

        for j in range(0, n - i - 1):
            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
                swapped = True

        if not swapped:
            break

    return arr


employee_salaries = [45000, 32000, 58000, 41000, 67000, 38000]

print("Original salaries:", employee_salaries)

sorted_salaries = bubble_sort(employee_salaries.copy())

print("Sorted salaries:", sorted_salaries)
```
