```python
products = {
    "Smartwatch": 8000,
    "Bluetooth Speaker": 3500,
    "Tablet": 25000,
    "Power Bank": 1200,
    "Earbuds": 4500
}

cart = {}


def display_products():
    print("\n- Available Products -")
    for product, price in products.items():
        print(product, "- ₹", price)


def add_product():
    display_products()
    product = input("\nEnter product name: ")

    if product in products:
        quantity = int(input("Enter quantity: "))

        if quantity <= 0:
            print("Quantity must be greater than 0.")
            return

        if product in cart:
            cart[product] += quantity
        else:
            cart[product] = quantity

        print(quantity, product, "added to cart.")
    else:
        print("Product not available.")


def remove_product():
    if not cart:
        print("\nCart is empty.")
        return

    display_cart()
    product = input("\nEnter product name to remove: ")

    if product in cart:
        quantity = int(input("Enter quantity to remove: "))

        if quantity >= cart[product]:
            del cart[product]
            print(product, "removed from cart.")
        else:
            cart[product] -= quantity
            print(quantity, product, "removed from cart.")
    else:
        print("Product not found in cart.")


def display_cart():
    print("\n- Shopping Cart -")

    if not cart:
        print("Cart is empty.")
        return

    print("Product\t\tPrice\tQuantity\tTotal")

    for product, quantity in cart.items():
        price = products[product]
        total = price * quantity
        print(product, "\t₹", price, "\t", quantity, "\t\t₹", total)


def calculate_subtotal():
    subtotal = 0
    for product, quantity in cart.items():
        subtotal += products[product] * quantity
    return subtotal


def apply_discount(subtotal):
    print("\nEnter discount percentage:")
    discount_percent = float(input())
    discount = subtotal * discount_percent / 100
    amount_after_discount = subtotal - discount
    return discount_percent, discount, amount_after_discount


def calculate_gst(amount_after_discount, gst_percent=18):
    gst = amount_after_discount * gst_percent / 100
    return gst


def generate_bill():
    if not cart:
        print("\nCart is empty. Cannot generate bill.")
        return

    subtotal = calculate_subtotal()
    discount_percent, discount, amount_after_discount = apply_discount(subtotal)
    gst = calculate_gst(amount_after_discount)
    final_amount = amount_after_discount + gst

    print("\n--")
    print("           FINAL BILL")
    print("--")
    print("Product\t\tQty\tPrice\tTotal")

    for product, quantity in cart.items():
        price = products[product]
        total = price * quantity
        print(product, "\t", quantity, "\t₹", price, "\t₹", total)

    print("--")
    print("Subtotal             : ₹", subtotal)
    print("Discount (", discount_percent, "%) : ₹", discount)
    print("Amount After Discount: ₹", amount_after_discount)
    print("GST (18%)            : ₹", gst)
    print("--")
    print("FINAL BILL           : ₹", final_amount)
    print("--")


def show_menu():
    print("\n--")
    print("       ONLINE SHOPPING CART SYSTEM")
    print("--")
    print("1. Display Products")
    print("2. Add Product")
    print("3. Remove Product")
    print("4. Display Cart")
    print("5. Generate Final Bill")
    print("6. Exit")
    print("--")


def main():
    while True:
        show_menu()
        choice = input("Enter your choice: ")

        if choice == "1":
            display_products()
        elif choice == "2":
            add_product()
        elif choice == "3":
            remove_product()
        elif choice == "4":
            display_cart()
        elif choice == "5":
            generate_bill()
        elif choice == "6":
            print("\nThank you for shopping!")
            break
        else:
            print("\nInvalid choice! Please try again.")


if __name__ == "__main__":
    main()
```
