```python
GST_RATE = 18
DISCOUNT_RATE = 10
DISCOUNT_LIMIT = 1000

cart = {}


def add_product():
    print("\n--- Add Product ---")

    name = input("Enter product name: ").strip()

    if not name:
        print("Invalid product name.")
        return

    try:
        price = float(input("Enter product price: "))
        quantity = int(input("Enter quantity: "))
    except ValueError:
        print("Please enter valid price and quantity.")
        return

    if price < 0 or quantity <= 0:
        print("Price must not be negative and quantity must be positive.")
        return

    if name in cart:
        cart[name]["quantity"] += quantity
    else:
        cart[name] = {
            "price": price,
            "quantity": quantity
        }

    print(f"{quantity} x {name} added to cart.")


def remove_product():
    print("\n--- Remove Product ---")

    if not cart:
        print("Cart is empty.")
        return

    name = input("Enter product name to remove: ").strip()

    if name in cart:
        del cart[name]
        print(f"{name} removed from cart.")
    else:
        print("Product not found in cart.")


def change_quantity():
    print("\n--- Change Quantity ---")

    if not cart:
        print("Cart is empty.")
        return

    name = input("Enter product name: ").strip()

    if name not in cart:
        print("Product not found in cart.")
        return

    try:
        quantity = int(input("Enter new quantity: "))
    except ValueError:
        print("Please enter a valid quantity.")
        return

    if quantity < 0:
        print("Quantity cannot be negative.")
        return

    if quantity == 0:
        del cart[name]
        print(f"{name} removed from cart.")
    else:
        cart[name]["quantity"] = quantity
        print(f"Quantity of {name} updated to {quantity}.")


def display_cart():
    print("\n--- Shopping Cart ---")

    if not cart:
        print("Cart is empty.")
        return

    print(f"{'Product':<20}{'Price':<12}{'Quantity':<10}{'Amount':<12}")
    print("-" * 54)

    for name, product in cart.items():
        amount = product["price"] * product["quantity"]

        print(
            f"{name:<20}"
            f"Rs.{product['price']:<9.2f}"
            f"{product['quantity']:<10}"
            f"Rs.{amount:.2f}"
        )


def calculate_subtotal():
    subtotal = 0

    for product in cart.values():
        subtotal += product["price"] * product["quantity"]

    return subtotal


def calculate_discount(subtotal):
    if subtotal >= DISCOUNT_LIMIT:
        return subtotal * DISCOUNT_RATE / 100

    return 0


def calculate_gst(amount):
    return amount * GST_RATE / 100


def display_final_bill():
    print("\n--- Final Bill ---")

    if not cart:
        print("Cart is empty.")
        return

    display_cart()

    subtotal = calculate_subtotal()
    discount = calculate_discount(subtotal)
    taxable_amount = subtotal - discount
    gst = calculate_gst(taxable_amount)
    final_amount = taxable_amount + gst

    print("\n--- Bill Summary ---")
    print(f"Subtotal       : Rs.{subtotal:.2f}")
    print(f"Discount       : Rs.{discount:.2f}")
    print(f"Taxable Amount : Rs.{taxable_amount:.2f}")
    print(f"GST ({GST_RATE}%)      : Rs.{gst:.2f}")
    print(f"Final Amount   : Rs.{final_amount:.2f}")


def main():
    while True:
        print("\n--- Online Shopping Cart System ---")
        print("1. Add Product")
        print("2. Remove Product")
        print("3. Change Quantity")
        print("4. Display Cart")
        print("5. Display Final Bill")
        print("6. Exit Program")

        choice = input("Enter choice: ").strip()

        if choice == "1":
            add_product()
        elif choice == "2":
            remove_product()
        elif choice == "3":
            change_quantity()
        elif choice == "4":
            display_cart()
        elif choice == "5":
            display_final_bill()
        elif choice == "6":
            print("Exiting system.")
            break
        else:
            print("Invalid choice, try again.")


if __name__ == "__main__":
    main()
```
