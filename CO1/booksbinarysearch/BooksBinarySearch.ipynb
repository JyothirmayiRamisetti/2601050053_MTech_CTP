```python
def Library(target):
    low = 0
    high = len(books) - 1

    while low <= high:
        mid = (low + high) // 2

        if books[mid] == target:
            return mid

        elif target < books[mid]:
            high = mid - 1

        else:
            low = mid + 1

    return -1


books = list(range(1, 1000001))

target = int(input("What book do you want to find: "))

result = Library(target)

if result != -1:
    print(f"Book at: {result}")
else:
    print("Book not found")
```
